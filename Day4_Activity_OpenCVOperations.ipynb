```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread(r"test.png")
```

```python
print(type(img))
```

```python
print(img.shape)
```

```python
if img is None:
    print("Error: Could not load image. Check the file path and name!")
else:
    # First argument must be the window name string, second is the image array
    print("showing image")
    cv2.imshow('Bootcamp Image', img) 
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
img_resize = cv2.resize(img, (250, 200), interpolation=cv2.INTER_AREA)
cv2.imshow('Bootcamp Image', img_resize)   #Resize
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_flip = cv2.flip(img,1)
cv2.imshow('Bootcamp Image', img_flip)     #Flip
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
img_crop=img[100:280,200:240]
cv2.imshow('Bootcamp Image', img_crop)     #Crop
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imwrite('cropped_image.jpg', img_crop) #Saving the image
```
