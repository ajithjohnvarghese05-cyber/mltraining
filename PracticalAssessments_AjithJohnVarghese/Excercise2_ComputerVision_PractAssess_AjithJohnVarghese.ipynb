```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the two images
img1 = cv2.imread('Excercise2_input_twigs.png')
img2 = cv2.imread('Excercise2_input_bricks.png')

# Check whether images are loaded successfully
if img1 is None or img2 is None:
    print("Error: Unable to load images. Check the file paths.")
else:
    print("Both images loaded successfully.")
```

```python
# Resize the second image to match the first image
img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

print("Image 1 dimensions:", img1.shape)
print("Image 2 dimensions:", img2.shape)
```

```python
# Blend the images
# 0.6 = weight of first image
# 0.4 = weight of second image
# 0 = gamma value

blended = cv2.addWeighted(img1, 0.5, img2, 0.5, 0)

print("Image blending completed successfully.")
```

```python
# Convert BGR images to RGB
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended, cv2.COLOR_BGR2RGB)

# Display all three images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(img1_rgb)
plt.title("Original Image 1")
plt.axis('off')

plt.subplot(1, 3, 2)
plt.imshow(img2_rgb)
plt.title("Original Image 2")
plt.axis('off')

plt.subplot(1, 3, 3)
plt.imshow(blended_rgb)
plt.title("Blended Image")
plt.axis('off')

plt.tight_layout()
plt.show()
```

```python
# Save all three images
cv2.imwrite('Excercise2_output_twigs.jpg', img1)
cv2.imwrite('Excercise2_output_bricks.jpg', img2)
cv2.imwrite('Excercise2_output_blended_image.jpg', blended)

print("All images saved successfully!")
```
