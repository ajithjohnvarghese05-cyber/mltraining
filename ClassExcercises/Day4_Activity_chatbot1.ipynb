```python
# Load FLAN-T5-small model
model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

# Use GPU if available; otherwise CPU
device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

print("Model loaded successfully!")
print("Device:", device)


# Few-shot example-based chatbot
def get_kalam_response(user_input):

    prompt = f"""
Answer in an inspirational, thoughtful style reflecting the values
and public teachings of Dr. APJ Abdul Kalam.

Q: Who is the Missile Man of India?
A: Dr. APJ Abdul Kalam is known as the Missile Man of India
for his significant contributions to India's missile programs.

Q: Who is Swami Vivekananda?
A: Swami Vivekananda was a spiritual leader who represented
Indian philosophy and values globally and inspired youth.

Q: Who is APJ Abdul Kalam?
A: Dr. APJ Abdul Kalam was the 11th President of India,
a brilliant scientist, and a beloved teacher known for
his vision of a developed India.

Q: {user_input}
A:
"""

    # Tokenize input
    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    ).to(device)

    # Generate response
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=False
        )

    # Decode response
    result = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return result.strip()


# CLI loop
print("\nWelcome to Dr. Kalam Chatbot!")
print("Type 'exit' to stop.\n")

while True:
    user_input = input("You: ")

    if user_input.lower() in ["exit", "quit"]:
        print("Chatbot: Thank you! Keep dreaming and keep working.")
        break

    reply = get_kalam_response(user_input)
    print("Chatbot:", reply)
```
