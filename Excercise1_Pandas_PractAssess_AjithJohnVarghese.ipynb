```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate random numerical values
np.random.seed(42)  # To ensure reproducible results

data = np.random.randint(10, 100, size=(10, 4))

# Create Pandas DataFrame
df = pd.DataFrame(data, columns=['f1', 'f2', 'f3', 'f4'])

# Display the DataFrame
print("Original DataFrame:")
display(df)
```

```python
# Create scatter plot
plt.figure(figsize=(8, 5))

plt.scatter(df['f2'], df['f3'], color='blue', marker='o')

# Add chart title and axis labels
plt.title('Scatter Plot: Relationship between f2 and f3')
plt.xlabel('f2')
plt.ylabel('f3')

# Display grid
plt.grid(True)

# Display the plot
plt.show()
```

```python
# Calculate row-wise mean across f1, f2, f3 and f4
df['f5'] = df[['f1', 'f2', 'f3', 'f4']].mean(axis=1)

# Display updated DataFrame
print("DataFrame with f5 (Row-wise Mean):")
display(df)
```

```python
# Export DataFrame to Excel
file_name = 'Excercise1_Output_Exploratory_Data_Analysis.xlsx'

df.to_excel(file_name, index=False)

print(f"Data successfully exported to {file_name}")
```

```python
# Read the exported Excel file
excel_data = pd.read_excel('Excercise1_Output_Exploratory_Data_Analysis.xlsx')

# Display the exported data
display(excel_data)
```
