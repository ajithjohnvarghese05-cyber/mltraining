```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta
```

```python
# --------------------------------------------------
# 1. Initialize random seeds
# --------------------------------------------------

np.random.seed(42)
random.seed(42)

# Number of records
N = 900
```

```python
# --------------------------------------------------
# 2. Define department-wise complaint master
# --------------------------------------------------

complaint_data = {

    "Pension Department": {
        "category": "Non-Receipt of Benefits",
        "complaints": [
            "I have not received my pension for the last three months despite submitting all required documents.",
            "The pension application has not been approved despite meeting the requirements.",
            "My monthly pension amount has not been credited to my bank account.",
            "I am facing difficulties in updating my pension details in government records.",
            "There is an unexplained delay in processing my retirement pension."
        ]
    },

    "Electricity Department": {
        "category": "Poor Service Quality",
        "complaints": [
            "There are frequent power interruptions in our locality.",
            "My electricity connection application has not been processed despite completing all formalities.",
            "The electricity meter installed at my residence is defective.",
            "Our village has been experiencing continuous power outages.",
            "The electricity bill contains incorrect charges."
        ]
    },

    "Water Resources Department": {
        "category": "Infrastructure Issues",
        "complaints": [
            "There is no proper drinking water supply in our village for the last two weeks.",
            "The drainage system is blocked and wastewater is overflowing onto the road.",
            "The public water pipeline is leaking and wasting drinking water.",
            "Our locality receives water supply only once in several days.",
            "The drinking water supplied to our village is contaminated."
        ]
    },

    "Public Works Department": {
        "category": "Infrastructure Issues",
        "complaints": [
            "The road in our area is damaged and has not been repaired despite repeated complaints.",
            "The street lights in our locality are not working, causing safety problems during night hours.",
            "A damaged bridge is creating difficulties for commuters.",
            "There are large potholes on the main road causing accidents.",
            "Construction work on the public road has remained incomplete for several months."
        ]
    },

    "Health Department": {
        "category": "Poor Service Quality",
        "complaints": [
            "I visited the government hospital several times but essential medicines are unavailable.",
            "The local public health centre lacks sufficient medical staff.",
            "Patients are waiting for several hours to receive treatment at the government hospital.",
            "The hospital does not have sufficient beds for admitted patients.",
            "Emergency medical services are not available in our locality."
        ]
    },

    "Education Department": {
        "category": "Administrative Issues",
        "complaints": [
            "The school building in our village requires urgent repairs and maintenance.",
            "The scholarship amount has not been credited to my bank account.",
            "There are insufficient teachers in the government school.",
            "My child's scholarship application has been pending for several months.",
            "Students are unable to access proper drinking water and sanitation facilities at school."
        ]
    },

    "Revenue Department": {
        "category": "Administrative Issues",
        "complaints": [
            "There is an unexplained delay in issuing my income certificate.",
            "My land ownership records contain incorrect information which needs correction.",
            "My application for a government certificate has been pending for several months.",
            "There is a delay in processing my land mutation application.",
            "The revenue office has not responded to my repeated requests."
        ]
    },

    "Social Welfare Department": {
        "category": "Non-Receipt of Benefits",
        "complaints": [
            "My application for a government welfare scheme has been pending for several months.",
            "I have not received the financial assistance promised under the government scheme.",
            "My welfare pension application has been rejected without providing a proper reason.",
            "I am eligible for a government assistance scheme but have not received the benefits.",
            "The financial assistance for eligible beneficiaries has been delayed."
        ]
    },

    "Transport Department": {
        "category": "Delay in Service",
        "complaints": [
            "Public transport buses are not operating regularly in our area.",
            "My driving licence application has not been processed.",
            "There is a delay in issuing my vehicle registration certificate.",
            "Government buses are overcrowded and not following scheduled timings.",
            "The regional transport office has not responded to my application."
        ]
    },

    "Police Department": {
        "category": "Public Safety",
        "complaints": [
            "I submitted a complaint about illegal activities but no action has been taken.",
            "There are frequent theft incidents in our residential area.",
            "My police complaint has not been registered despite repeated requests.",
            "There is no adequate police patrolling in our locality.",
            "I have reported a public safety issue but have not received any response."
        ]
    }
}

# --------------------------------------------------
# 3. Verify complaint master
# --------------------------------------------------

total_narratives = sum(
    len(details["complaints"])
    for details in complaint_data.values()
)

assert len(complaint_data) == 10
assert total_narratives == 50

print("Department count:", len(complaint_data))
print("Unique complaint narratives:", total_narratives)
```

```python
# --------------------------------------------------
# 4. Define other dataset attributes
# --------------------------------------------------

states = [
    "Kerala",
    "Tamil Nadu",
    "Karnataka",
    "Andhra Pradesh",
    "Telangana",
    "Maharashtra",
    "Delhi",
    "Gujarat",
    "Rajasthan",
    "Uttar Pradesh"
]

statuses = [
    "Pending",
    "In Progress",
    "Resolved",
    "Closed",
    "Escalated"
]

# Assign equal probability to each status
status_probabilities = [0.20, 0.20, 0.35, 0.15, 0.10]

# Date range
start_date_range = datetime(2025, 1, 1)
end_date_range = datetime(2026, 11, 24)
```

```python
# --------------------------------------------------
# 5. Function to generate random complaint date
# --------------------------------------------------

def random_start_date():

    total_days = (
        end_date_range - start_date_range
    ).days

    random_days = random.randint(0, total_days)

    return start_date_range + timedelta(days=random_days)
```

```python
# --------------------------------------------------
# 6. Generate N grievance records
# --------------------------------------------------

records = []

department_names = list(complaint_data.keys())

for i in range(N):

    # Select department
    department = random.choice(department_names)

    # Select corresponding category
    category = complaint_data[department]["category"]

    # Select complaint narrative from department master
    complaint = random.choice(
        complaint_data[department]["complaints"]
    )

    # Select state
    state = random.choice(states)

    # Select complaint status
    status = random.choices(
        statuses,
        weights=status_probabilities,
        k=1
    )[0]

    # Generate complaint registration date
    start_date = random_start_date()

    # Generate resolution date based on status
    if status in ["Resolved", "Closed"]:

        duration = random.randint(2, 90)

        end_date = start_date + timedelta(days=duration)

        # Ensure end date does not exceed the dataset cutoff
        if end_date > end_date_range:
            end_date = end_date_range

    else:
        end_date = pd.NaT

    # Generate unique complaint ID
    complaint_id = f"JSM{100001 + i}"

    # Add record
    records.append({

        "Department": department,

        "Status": status,

        "Citizen Complaint Narrative": complaint,

        "Complaint ID": complaint_id,

        "State": state,

        "category": category,

        "start date": start_date,

        "end date": end_date

    })
```

```python
# --------------------------------------------------
# 7. Create Pandas DataFrame
# --------------------------------------------------

df = pd.DataFrame(records)

# Convert date columns
df["start date"] = pd.to_datetime(df["start date"])
df["end date"] = pd.to_datetime(df["end date"])
```

```python
# --------------------------------------------------
# 8. Calculate resolution time
# --------------------------------------------------

df["Resolution Days"] = (
    df["end date"] - df["start date"]
).dt.days
```

```python
# --------------------------------------------------
# 9. Dataset validation
# --------------------------------------------------

print("\n========== DATASET SUMMARY ==========")

print("Total Records:", len(df))
print("Total Columns:", len(df.columns))
print("Unique Complaint IDs:", df["Complaint ID"].nunique())
print("Unique Narratives:", df["Citizen Complaint Narrative"].nunique())
print("Total Departments:", df["Department"].nunique())

assert len(df) == N
assert df["Complaint ID"].nunique() == N
assert df["Citizen Complaint Narrative"].nunique() == 50
```
