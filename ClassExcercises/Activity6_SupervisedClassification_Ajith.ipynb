```python
#Supervised Regression
from sklearn.linear_model import LogisticRegression
```

```python
# 1. Input Features: 
X_applns = [
    [2.5, 1], [3.0,1], [5.0,1], [12.0,1], [2.5,0], [3.0,0], [8.0,0], [12.0,0]]  # Income, Docs Available

# Target Labels: 
#1 = Approved, 0 = Rejected
y_permit = [1, 1, 1, 0, 0, 0, 0, 0]             # Decision

# 2. Train the Classifier
classifier = LogisticRegression()
classifier.fit(X_applns, y_permit)

# 3. Predict for a new applicant: ₹3.5 Lakhs income, complete documents [3.5, 1]
new_applicant = [[3.0,0]]
result = classifier.predict(new_applicant)

status = "Approved (1)" if result[0] == 1 else "Rejected (0)"

print(f"Application Result: {status}")
```
