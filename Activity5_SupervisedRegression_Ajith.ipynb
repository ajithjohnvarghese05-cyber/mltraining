```python
import matplotlib.pyplot as plt
```

```python
#Supervised Regression
from sklearn.linear_model import LinearRegression
# 1. Input Data (Must be 2D array for sklearn, Target is 1D)
X_pages = [[10], [20], [30], [40]]  # Page counts
y_days = [2, 5, 8, 12]             # Actual processing days

# 2. Train the Model
model = LinearRegression()
model.fit(X_pages, y_days)

# 3. Predict for a new file with 50 pages
new_file = [[25]]
predicted_days = model.predict(new_file)

print(f"Predicted days for {new_file} pages: {predicted_days[0]:.1f} days")

plt.plot(X_pages,y_days, marker="x")
plt.title("My Plot X=Pages Y=DaysReqd")
plt.xlabel("Pages")
plt.ylabel("Days Required")
plt.show()
```
